# Baseline: generic object detector on a capital photo

This baseline uses pretrained COCO YOLOv8n; COCO labels have no dedicated Doric/Ionic/Corinthian capital class. This is a qualitative illustration, not a comparable 3-class benchmark. The photo is by George E. Koronaios, [CC BY-SA 4.0](https://commons.wikimedia.org/wiki/File:Ionic_capital_(Church_of_St_Catherine,_Plaka)_on_February_25,_2022.jpg).

```
%pip install -q ultralytics==8.4.163
import requests
from pathlib import Path
from ultralytics import YOLO
name="File:Ionic capital (Church of St Catherine, Plaka) on February 25, 2022.jpg"
s=requests.Session(); s.headers.update({"User-Agent":"SilvanaClassicalCapitalsResearch/1.0 (educational project)"})
r=s.get("https://commons.wikimedia.org/w/api.php",params={"action":"query","format":"json","prop":"imageinfo","iiprop":"url","iiurlwidth":1200,"titles":name},timeout=30); r.raise_for_status()
page=next(iter(r.json()["query"]["pages"].values()))
u=page["imageinfo"][0].get("thumburl",page["imageinfo"][0]["url"])
img=Path('/content/baseline_ionic.jpg'); rr=s.get(u,timeout=60); rr.raise_for_status(); img.write_bytes(rr.content)
model=YOLO('yolov8n.pt')
result=model.predict(str(img),conf=0.25,save=True,project='/content/baseline',name='predictions',exist_ok=True)
print('Generic COCO detections:',[(model.names[int(b.cls.item())],round(float(b.conf.item()),2)) for b in result[0].boxes])
```
